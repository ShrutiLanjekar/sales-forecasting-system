```python
import pandas as pd

df = pd.read_csv("../data/processed/feature_engineered.csv")
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
df = df.sort_values(['State', 'Date'])
```

```python
target = 'Sales'
```

```python
features = [
    'lag_1',
    'lag_7',
    'lag_30',
    'rolling_mean_7',
    'rolling_mean_30',
    'rolling_std_7',
    'day_of_week',
    'month',
    'quarter',
    'week_of_year',
    'year',
    'is_holiday'
]
```

```python
forecast_horizon = 56

train_df = df.iloc[:-forecast_horizon]

valid_df = df.iloc[-forecast_horizon:]
```

```python
X_train = train_df[features]

y_train = train_df[target]
```

```python
X_valid = valid_df[features]

y_valid = valid_df[target]
```

```python
print(X_train.shape)

print(X_valid.shape)

print(y_train.shape)

print(y_valid.shape)
```

```python
print(train_df['Date'].min())

print(train_df['Date'].max())

print(valid_df['Date'].min())

print(valid_df['Date'].max())
```

```python
train_df.to_csv(
    "../data/processed/train_data.csv",
    index=False
)

valid_df.to_csv(
    "../data/processed/valid_data.csv",
    index=False
)
```
