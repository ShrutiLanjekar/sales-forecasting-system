```python
import pandas as pd

import matplotlib.pyplot as plt

import joblib
```

```python
sarima_mae = 215483378.08
sarima_rmse = 246426700.29

prophet_mae = 54073398.30
prophet_rmse = 72618561.96

xgb_mae = 546603.31
xgb_rmse =686993.74

lstm_mae = 4314511.34
lstm_rmse = 10187611.82
```

```python
results_df = pd.DataFrame({
    'Model': [
        'SARIMA',
        'Prophet',
        'XGBoost',
        'LSTM'
    ],
    
    'MAE': [
        sarima_mae,
        prophet_mae,
        xgb_mae,
        lstm_mae
    ],
    
    'RMSE': [
        sarima_rmse,
        prophet_rmse,
        xgb_rmse,
        lstm_rmse
    ]
})
```

```python
print(results_df)
```

```python
results_df = results_df.sort_values(
    by='RMSE'
)
```

```python
best_model_row = results_df.iloc[0]

best_model_name = best_model_row['Model']

print(best_model_name)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(
    results_df['Model'],
    results_df['RMSE']
)

plt.title("Model RMSE Comparison")

plt.ylabel("RMSE")

plt.show()
```

```python
results_df.to_csv(
    "../outputs/model_comparison.csv",
    index=False
)
```

```python
print(f"Best Model: {best_model_name}")
```
