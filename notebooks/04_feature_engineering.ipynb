```python
import pandas as pd

df = pd.read_csv("../data/processed/final_processed.csv")

df['Date'] = pd.to_datetime(df['Date'])

df.head()
```

```python
df = df.sort_values(['State', 'Date'])
```

```python
df['lag_1'] = (
    df.groupby('State')['Sales']
    .shift(1)
)
```

```python
df['lag_7'] = (
    df.groupby('State')['Sales']
    .shift(7)
)
```

```python
df['lag_30'] = (
    df.groupby('State')['Sales']
    .shift(30)
)
```

Create Rolling Mean

```python
df['rolling_mean_7'] = (
    df.groupby('State')['Sales']
    .transform(lambda x: x.rolling(7).mean())
)
```

```python
df['rolling_mean_30'] = (
    df.groupby('State')['Sales']
    .transform(lambda x: x.rolling(30).mean())
)
```

Rolling Standard Deviation

```python
df['rolling_std_7'] = (
    df.groupby('State')['Sales']
    .transform(lambda x: x.rolling(7).std())
)
```

Day Features

```python
df['day_of_week'] = df['Date'].dt.dayofweek
```

```python
df['month'] = df['Date'].dt.month
```

```python
df['quarter'] = df['Date'].dt.quarter
```

```python
df['week_of_year'] = (
    df['Date']
    .dt.isocalendar()
    .week
    .astype(int)
)
```

```python
df['year'] = df['Date'].dt.year
```

Holiday Feature

```python
import holidays
us_holidays = holidays.US()

df['is_holiday'] = (
    df['Date']
    .isin(us_holidays)
    .astype(int)
)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```
